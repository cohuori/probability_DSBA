# Method of Moments

Suppose that $X_1,\ldots,X_n$ is a random sample from a parametric family of probability distributions $\mathcal P=\left\{P_\theta:\theta\in\Theta\right\}$, where

- $\Theta\subseteq\mathbb R^r$ is the **parameter space**, and
- $\theta=(\theta_1,\ldots,\theta_r)^T$ is the unknown parameter vector.

The objective of parametric estimation is to construct a statistic

$$
\widehat\theta
=
\widehat\theta(X_1,\ldots,X_n)
$$

that provides a good approximation of the true parameter value.

One of the oldest and most intuitive estimation procedures is the **Method of Moments (MOM)**, introduced by **Karl Pearson (1894)**. Its central idea is extremely simple:

> Replace unknown theoretical moments by their corresponding sample moments and solve the resulting system of equations for the unknown parameters.

Although modern statistics often favors Maximum Likelihood Estimation (MLE), the method of moments remains an important estimation technique due to its simplicity, computational efficiency, and broad applicability.

## Population Moments

The method of moments is based on the observation that many probability distributions are completely determined by their moments. Let $X$ be a random variable.

- The **$k$-th raw (non-central) moment** of $X$ is defined by

$$
\mu_k'
=
\mathbb E[X^k],
\qquad
k=1,2,\ldots.
$$

- The **$k$-th central moment** is defined by

$$
\mu_k
=
\mathbb E[(X-\mu)^k].
$$


The first central moments are

$$
\begin{aligned}
\mu_1 &=0,\\
\mu_2 &=\operatorname{Var}(X),\\
\mu_3 &\text{ measures the asymmetry (skewness) of the distribution},\\
\mu_4 &\text{ is related to the kurtosis of the distribution}.
\end{aligned}
$$

Although central moments describe many important properties of a distribution, the Method of Moments is usually formulated in terms of **raw moments**, since they lead to simpler systems of equations.

## Moments as Functions of the Parameters

For a parametric family $\mathcal P=\{P_\theta:\theta\in\Theta\}$, each theoretical moment is a function of the unknown parameter, $\mu_k'(\theta)=
\mathbb E_\theta[X^k]$.

For example, if $X\sim N(\mu,\sigma^2)$, then

$$
\begin{aligned}
\mathbb E[X]
&=\mu,
\\[1ex]
\mathbb E[X^2]
&=\mu^2+\sigma^2.
\end{aligned}
$$

Consequently, if the theoretical moments determine the unknown parameters uniquely, they may be used to estimate those parameters.

## Sample Moments

Since the population moments are unknown, they must be estimated using the observed sample.

### Definition

Let $X_1,\ldots,X_n$ be a random sample. The **$k$-th sample raw moment** is defined by

$$
\boxed{
M_k'
=
\frac1n
\sum_{i=1}^n
X_i^k.
}
$$

In particular, $M_1'=\overline{X}$. Each sample moment is a statistic and therefore a random variable.

### Why Sample Moments?

The fundamental justification for the Method of Moments follows directly from the Law of Large Numbers. Indeed, provided the $k$-th moment exists,

$$
\frac1n
\sum_{i=1}^n
X_i^k
\xrightarrow{P}
\mathbb E[X^k].
$$


Therefore, for sufficiently large samples,

$$
M_k'
\approx
\mu_k'.
$$

This observation motivates replacing unknown population moments by their empirical counterparts.

## General Method of Moments

Let $g_1,\ldots,g_r:\mathbb R\rightarrow\mathbb R$, be measurable functions such that $\mathbb E_\theta[g_i(X)]$ exists for every parameter value. Here $\theta \in \mathbb R^r$.

Define the theoretical quantities

$$
m_i(\theta)
:=
\mathbb E_\theta[g_i(X)],
\qquad
i=1,\ldots,r.
$$

The corresponding empirical quantities are

$$
\widehat m_i
:=
\frac1n
\sum_{j=1}^n
g_i(X_j).
$$

The Method of Moments estimates the parameter by solving the system

$$
\boxed{
m_i(\theta)
=
\widehat m_i,
\qquad
i=1,\ldots,r.
}
$$

If this system admits a unique solution $\widehat\theta$, then $\widehat\theta$ is called the **Method of Moments estimator (MME)**.

### Remark

The functions $g_1,\ldots,g_r$ are called the **moment functions** (or **estimating functions**).

The classical Method of Moments corresponds to the particular choice

$$
g_i(x)=x^i,
\qquad
i=1,\ldots,r.
$$

More general choices of the functions $g_i$ lead to generalized moment-based estimation procedures, such as the **Generalized Method of Moments (GMM)** used extensively in econometrics.

##  Algorithm of Classical Method of Moments

Let $\theta=(\theta_1,\ldots,\theta_r)$, be an unknown parameter vector. The estimator is obtained as follows:

1. Compute the first $r$ theoretical moments

$$
\mathbb E_\theta[X],
\;
\mathbb E_\theta[X^2],
\;
\ldots,
\;
\mathbb E_\theta[X^r].
$$

2. Compute the corresponding sample moments

$$
M_1',
\;
M_2',
\;
\ldots,
\;
M_r'.
$$

3. Solve the system

$$
\mathbb E_\theta[X^k]
=
M_k',
\qquad
k=1,\ldots,r.
$$

4. The solution is the Method of Moments estimator

$$
\widehat\theta_{\mathrm{MOM}}.
$$

For many standard probability distributions, this system admits an explicit analytical solution. In more complicated models, numerical methods may be required.

### Example: Normal Distribution

Suppose $X_1,\ldots,X_n \overset{\mathrm{iid}}{\sim} N(\mu,\sigma^2)$, where both parameters $\mu$ and $\sigma^2$ are unknown.

Since there are two unknown parameters, we equate the first two theoretical moments to the corresponding sample moments.

The first theoretical moment is $\mathbb E[X]=\mu$, while the second theoretical moment is $\mathbb E[X^2]=\mu^2+\sigma^2$.

The Method of Moments equations are therefore

$$
\mu
=
\overline X,
\qquad
\mu^2+\sigma^2
=
\frac1n
\sum_{i=1}^{n}
X_i^2.
$$

We obtain the Method of Moments estimators

$$
\boxed{
\widehat{\mu}_{\mathrm{MOM}}
=
\overline X,
\qquad
\widehat{\sigma}^2_{\mathrm{MOM}}
=
\frac1n
\sum_{i=1}^{n}
(X_i-\overline X)^2.
}
$$

Notice that the estimator of the variance is **biased**, since it uses the denominator $n$ rather than $n-1$. Nevertheless, it is **consistent** and coincides with the Maximum Likelihood Estimator for the variance of a normal distribution.

# Asymptotic Properties of the Method of Moments

Like most estimation procedures, the Method of Moments is primarily justified through its **large-sample properties**.

Under suitable regularity conditions, Method of Moments estimators are

- consistent,
- asymptotically normal,
- and often straightforward to compute.

However, unlike Maximum Likelihood Estimators, they are generally **not asymptotically efficient**.

## Consistency of the Method of Moments

Suppose that $X_1,\ldots,X_n$ is a random sample from a distribution depending on the unknown parameter $\theta\in\Theta$.

Assume that the required population moments exist, and also, that the mapping

$$
\theta
\longmapsto
\left(
\mu_1'(\theta),
\ldots,
\mu_r'(\theta)
\right)
$$

is continuous, and it is one-to-one (identifiable).

Then the Method of Moments estimator satisfies

$$
\boxed{
\widehat{\theta}_{\mathrm{MOM}}
\xrightarrow{P}
\theta_0.
}
$$

## Asymptotic Normality

The consistency theorem may be strengthened considerably.

Assume that, additionally to the previous conditions, the inverse mapping $m^{-1}$ can be extended to a function defined on all of $\mathbb R^k$, and is continuously differentiable at every point of the set $m(\Theta)$. Moreover, for every $i=1,\ldots,k$ and every $\theta\in\Theta$, the moment function satisfies $0<\operatorname{Var}_\theta\!\left(g_i(X)\right)<+\infty$. Then,

$$
\boxed{
\sqrt n
\left(
\widehat{\theta}_{\mathrm{MOM}}
-
\theta_0
\right)
\xrightarrow{d}
N
\left(
0,
\Sigma_{\mathrm{MOM}}
\right).
}
$$

Consequently, for sufficiently large samples,

$$
\widehat{\theta}_{\mathrm{MOM}}
\approx
N
\left(
\theta_0,
\frac1n
\Sigma_{\mathrm{MOM}}
\right).
$$

## Advantages of the Method of Moments

- It is conceptually simple.

- It often leads to explicit closed-form estimators.

- It usually requires only elementary algebra.

- It does not require optimization algorithms.

- It is applicable even when the likelihood function is difficult to maximize.

For these reasons, the Method of Moments is frequently used to obtain initial parameter estimates for more sophisticated estimation procedures.

## Limitations of the Method of Moments

Despite its simplicity, the Method of Moments has several important drawbacks.

- It generally ignores much of the information contained in the probability distribution.

- Different choices of moments may produce different estimators.

- Method of Moments estimators are not usually efficient.

- They may have larger variance than maximum likelihood estimators.

- In some models, the resulting system of equations may not admit a unique solution.

Consequently, although the Method of Moments is often computationally convenient, it is generally statistically less efficient than Maximum Likelihood Estimation.

# Maximum Likelihood Estimation

The Method of Moments estimates the unknown parameters by matching theoretical and empirical moments.

An alternative approach is to ask a fundamentally different question:

> **Which parameter value makes the observed sample the most plausible?**

This idea leads to one of the most important estimation procedures in statistics: **Maximum Likelihood Estimation (MLE)**.

Introduced by **Ronald A. Fisher** in the early twentieth century, maximum likelihood estimation has become the standard method for estimating parameters in statistics, econometrics, machine learning, and many other areas of data science.

Unlike the Method of Moments, which uses only a few characteristics of the distribution (its moments), maximum likelihood estimation exploits the **entire probability model** and therefore usually produces more efficient estimators.

## The Likelihood Function

Suppose first that $X$ has probability density function $f(x;\theta)$. Consider the sample $X_1,\ldots,X_n \sim X$.

The **likelihood function** is a function of $\theta$, defined as

$$
L(\theta)
=
L(\theta;x_1,\ldots,x_n)
=
\prod_{i=1}^{n}
f(x_i;\theta).
$$

### Discrete Case

If the observations come from a discrete distribution with probability mass function $p(x;\theta)$, the likelihood function becomes

$$
L(\theta)
=
\prod_{i=1}^{n}
p(x_i;\theta).
$$

Therefore, the definition of the likelihood is identical for continuous and discrete distributions; only the underlying probability function changes.

### Interpretation of the Likelihood

The likelihood function should not be interpreted as the probability that the parameter equals a particular value. The parameter is **not** considered to be a random variable.

Instead, $L(\theta)$ answers the following question:

> *Assuming that the parameter equals $\theta$, how compatible is the observed sample with the statistical model?*

Consequently,

- larger values of $L(\theta)$ indicate that the observed data are more compatible with the model,

- smaller values indicate that the observed sample would be less likely under that parameter value.

The goal of maximum likelihood estimation is therefore to choose the parameter that maximizes the likelihood.

## The Log-Likelihood Function

The **log-likelihood function** is defined by

$$
\ell(\theta)
=
\log L(\theta).
$$

Using the identity $\log(ab)=\log a+\log b$, the product becomes a sum,

$$
\boxed{
\ell(\theta)
=
\sum_{i=1}^{n}
\log f(x_i;\theta).
}
$$

This transformation greatly simplifies differentiation and optimization. Notice that maximizing the likelihood and maximizing the log-likelihood always produce exactly the same estimator, since

$$
\operatorname*{arg\,max}_{\theta}
L(\theta)
=
\operatorname*{arg\,max}_{\theta}
\ell(\theta).
$$

## Definition of the Maximum Likelihood Estimator

The **Maximum Likelihood Estimator (MLE)** is the parameter value that maximizes the likelihood function.

Formally, a statistic $\widehat\theta$ is called a **maximum likelihood estimator** if

$$
\boxed{
\widehat\theta
=
\operatorname*{arg\,max}_{\theta\in\Theta}
L(\theta).
}
$$

Equivalently, $\widehat\theta=\operatorname*{arg\,max}_{\theta\in\Theta} \ell(\theta)$.

The observed numerical value of $\widehat\theta$ is called the **maximum likelihood estimate**.

### Computing the MLE

When the likelihood is differentiable, the maximum likelihood estimator is typically obtained in three steps.

1. Construct the likelihood function $L(\theta)$.

2. Compute the log-likelihood $\ell(\theta)$.

3. Differentiate and solve $\frac{\partial\ell(\theta)}{\partial\theta}=0$.

Finally, verify that the solution corresponds to a maximum by checking the second derivative or using other optimization arguments. For multi-parameter models, the same procedure is applied using partial derivatives with respect to each parameter.

## Example: MLE for a Normal Distribution

Suppose $X_1,\ldots,X_n\overset{\mathrm{iid}}{\sim} N(\mu,\sigma^2)$,  where both parameters $\mu$ and $\sigma^2$ are unknown.

The probability density function is

$$
f(x;\mu,\sigma^2)
=
\frac{1}{\sqrt{2\pi\sigma^2}}
\exp\left(
-
\frac{(x-\mu)^2}{2\sigma^2}
\right).
$$

The likelihood function is

$$
L(\mu,\sigma^2)
=
\left(
\frac{1}{\sqrt{2\pi\sigma^2}}
\right)^n
\exp\left(
-
\frac{1}{2\sigma^2}
\sum_{i=1}^{n}
(X_i-\mu)^2
\right).
$$

Taking logarithms,

$$
\begin{aligned}
\ell(\mu,\sigma^2)
&=
\log L(\mu,\sigma^2)
\\[1ex]
&=
-\frac{n}{2}\log(2\pi)
-\frac{n}{2}\log(\sigma^2)
-\frac{1}{2\sigma^2}
\sum_{i=1}^{n}
(X_i-\mu)^2.
\end{aligned}
$$

---

### Step 1 — Estimate of the Mean

Differentiate the log-likelihood with respect to $\mu$:

$$
\frac{\partial\ell}{\partial\mu}
=
\frac{1}{\sigma^2}
\sum_{i=1}^{n}
(X_i-\mu).
$$

Setting the derivative equal to zero, which gives

$$
n\mu
=
\sum_{i=1}^{n}
X_i.
$$

Hence,

$$
\boxed{
\widehat{\mu}_{\mathrm{MLE}}
=
\overline X
}
$$

---

### Step 2 — Estimate of the Variance

Differentiate the log-likelihood with respect to $\sigma^2$:

$$
\frac{\partial\ell}{\partial\sigma^2}
=
-\frac{n}{2\sigma^2}
+
\frac{1}{2(\sigma^2)^2}
\sum_{i=1}^{n}
(X_i-\mu)^2.
$$

Setting this derivative equal to zero,

$$
-\frac{n}{2\sigma^2}
+
\frac{1}{2(\sigma^2)^2}
\sum_{i=1}^{n}
(X_i-\mu)^2
=
0.
$$

Therefore,

$$
\sigma^2
=
\frac1n
\sum_{i=1}^{n}
(X_i-\mu)^2.
$$

Substituting $\mu=\widehat{\mu}_{\mathrm{MLE}}=\overline X$, we obtain

$$
\boxed{
\widehat{\sigma}^2_{\mathrm{MLE}}
=
\frac1n
\sum_{i=1}^{n}
(X_i-\overline X)^2
}
$$



Notice that $\widehat{\sigma}^2_{\mathrm{MLE}}$ is a biased estimator of the population variance. Nevertheless, it is **consistent**, **asymptotically normal**, and **asymptotically efficient**. Moreover, for the normal distribution, it coincides with the Method of Moments estimator of the variance.

# The Score Function

Suppose that the log-likelihood is differentiable. The **score function** is defined by

$$
\boxed{
U(\theta)
=
\frac{\partial}{\partial\theta}
\ell(\theta).
}
$$

If the parameter is a vector,

$$
\theta=(\theta_1,\ldots,\theta_p),
$$

then the score function is the gradient of the log-likelihood,

$$
\boxed{
U(\theta)
=
\nabla_\theta
\ell(\theta)
=
\left(
\frac{\partial\ell}{\partial\theta_1},
\ldots,
\frac{\partial\ell}{\partial\theta_p}
\right)^T.
}
$$

Thus, the score measures how rapidly the log-likelihood changes as the parameter varies.

## Maximum Likelihood Equations

Since the maximum of a differentiable function is attained at a critical point, the maximum likelihood estimator usually satisfies

$$
\boxed{
U(\widehat\theta)=0.
}
$$

These equations are called the **likelihood equations** or **score equations**.

For a single parameter,

$$
\frac{\partial\ell(\theta)}
{\partial\theta}
=
0,
$$

whereas for several parameters,

$$
\frac{\partial\ell(\theta)}
{\partial\theta_j}
=
0,
\qquad
j=1,\ldots,p.
$$

The solution of this system is the maximum likelihood estimator, provided that it corresponds to a maximum of the likelihood.

## Interpretation

The score function plays a role analogous to the derivative of an ordinary function.

- If

$$
U(\theta)>0,
$$

the log-likelihood is increasing.

- If

$$
U(\theta)<0,
$$

the log-likelihood is decreasing.

- If

$$
U(\theta)=0,
$$

the log-likelihood has a stationary point, which may correspond to a local maximum, a local minimum, or a saddle point.

Consequently, solving

$$
U(\theta)=0
$$

is the first step in computing the maximum likelihood estimator.

# A Fundamental Property of the Score Function

One of the most important theoretical properties of the score function is that its expectation is zero.

This result forms the basis for many asymptotic properties of maximum likelihood estimators and plays a central role in the derivation of the Fisher Information.

---

## Theorem

Suppose that the regularity conditions allowing differentiation under the integral sign are satisfied.

Then,

$$
\boxed{
\mathbb E_\theta
\!\left[
U(\theta)
\right]
=
0.
}
$$

In other words, the score function is an unbiased estimating function.

---

## Proof

Recall that

$$
U(\theta)
=
\frac{\partial}{\partial\theta}
\ell(\theta)
=
\frac{\partial}{\partial\theta}
\log L(\theta).
$$

For a random sample,

$$
L(\theta)
=
\prod_{i=1}^{n}
f(X_i;\theta),
$$

and therefore

$$
U(\theta)
=
\sum_{i=1}^{n}
\frac{\partial}{\partial\theta}
\log f(X_i;\theta).
$$

Hence,

$$
\mathbb E_\theta[U(\theta)]
=
n
\,
\mathbb E_\theta
\!\left[
\frac{\partial}{\partial\theta}
\log f(X;\theta)
\right].
$$

Since

$$
\frac{\partial}{\partial\theta}
\log f(x;\theta)
=
\frac{1}{f(x;\theta)}
\frac{\partial f(x;\theta)}
{\partial\theta},
$$

we obtain

$$
\begin{aligned}
\mathbb E_\theta
\!\left[
\frac{\partial}{\partial\theta}
\log f(X;\theta)
\right]
&=
\int
\frac{1}{f(x;\theta)}
\frac{\partial f(x;\theta)}
{\partial\theta}
f(x;\theta)
\,dx
\\[1ex]
&=
\int
\frac{\partial f(x;\theta)}
{\partial\theta}
\,dx
\\[1ex]
&=
\frac{\partial}{\partial\theta}
\int
f(x;\theta)
\,dx
\\[1ex]
&=
\frac{\partial}{\partial\theta}(1)
\\[1ex]
&=
0.
\end{aligned}
$$

Therefore,

$$
\boxed{
\mathbb E_\theta
\!\left[
U(\theta)
\right]
=
0.
}
$$

---

## Remark

The proof relies on the ability to interchange differentiation and integration,

$$
\frac{\partial}{\partial\theta}
\int
f(x;\theta)\,dx
=
\int
\frac{\partial f(x;\theta)}
{\partial\theta}
\,dx,
$$

which is guaranteed under suitable regularity conditions.

These assumptions are satisfied for most classical parametric families encountered in practice.

## Corollary

Since

$$
\mathbb E_\theta[U(\theta)]
=
0,
$$

the variance of the score function is

$$
\operatorname{Var}(U(\theta))
=
\mathbb E_\theta
\!\left[
U(\theta)^2
\right].
$$

Consequently,

$$
\boxed{
I(\theta)
=
\operatorname{Var}(U(\theta))
=
\mathbb E_\theta
\!\left[
U(\theta)^2
\right],
}
$$

which is known as the **Information Equality**.

# Observed Fisher Information

Finding a critical point is not sufficient to guarantee that the likelihood is maximized.

The curvature of the log-likelihood determines whether the stationary point is a maximum or a minimum.

This leads to the notion of **observed Fisher information**.

## Definition

For a one-dimensional parameter,

$$
\boxed{
J(\theta)
=
-
\frac{\partial^2\ell(\theta)}
{\partial\theta^2}.
}
$$

The quantity

$$
J(\theta)
$$

is called the **observed Fisher information**.

For vector parameters,

$$
\boxed{
J(\theta)
=
-
\nabla_\theta^2
\ell(\theta),
}
$$

that is, the negative Hessian matrix of the log-likelihood.

## Interpretation

The observed Fisher information measures the local curvature of the log-likelihood.

A sharply peaked likelihood has large curvature and therefore contains more information about the unknown parameter.

Conversely, a flat likelihood changes very little as the parameter varies, making estimation more difficult.

Consequently,

- large Fisher information corresponds to more precise estimation,

- small Fisher information corresponds to greater uncertainty.

# Fisher Information

Since the observed Fisher information depends on the particular sample, it is itself a random variable.

A more fundamental quantity is obtained by taking its expectation.

## Definition

The **Fisher information** is

$$
\boxed{
I(\theta)
=
\mathbb E_\theta
\left[
J(\theta)
\right]
=
-
\mathbb E_\theta
\left[
\frac{\partial^2\ell(\theta)}
{\partial\theta^2}
\right].
}
$$

For vector parameters,

$$
\boxed{
I(\theta)
=
-
\mathbb E_\theta
\left[
\nabla_\theta^2
\ell(\theta)
\right].
}
$$

Unlike the observed information,

$$
J(\theta),
$$

the Fisher information depends only on the statistical model and not on the observed sample.

## Information Equality

Under suitable regularity conditions,

$$
\boxed{
I(\theta)
=
\mathbb E_\theta
\left[
U(\theta)^2
\right].
}
$$

Equivalently,

$$
\boxed{
I(\theta)
=
\operatorname{Var}
\bigl(
U(\theta)
\bigr),
}
$$

since

$$
\mathbb E_\theta
\left[
U(\theta)
\right]
=
0.
$$

Thus, the Fisher information may be computed either from the expected curvature of the log-likelihood or from the variance of the score function.

## Why Is Fisher Information Important?

Fisher information quantifies the amount of information that the observed sample contains about the unknown parameter.

It plays a central role throughout mathematical statistics.

In particular,

- it determines the asymptotic variance of the maximum likelihood estimator,

- it appears in the Cramér–Rao Lower Bound,

- it characterizes asymptotic efficiency,

- and it provides the basis for likelihood ratio, score, and Wald hypothesis tests.

For these reasons, Fisher information is one of the most fundamental concepts in statistical inference.

## Summary

The differential quantities associated with the likelihood are

| Quantity | Definition | Purpose |
|----------|------------|---------|
| Score | $$U(\theta)=\dfrac{\partial\ell}{\partial\theta}$$ | Finds stationary points |
| Observed Fisher Information | $$J(\theta)=-\dfrac{\partial^2\ell}{\partial\theta^2}$$ | Measures local curvature |
| Fisher Information | $$I(\theta)=\mathbb E[J(\theta)]$$ | Measures information contained in the model |

Together, these quantities form the mathematical foundation of maximum likelihood estimation.

# Fundamental Properties of Maximum Likelihood Estimators

Maximum likelihood estimators possess several important large-sample properties.

Under suitable regularity conditions, the MLE is

- consistent,
- asymptotically normal,
- asymptotically efficient,
- and invariant under transformations.

These properties explain why maximum likelihood estimation plays such a central role in modern statistical inference.

It is important to emphasize that these results are generally **asymptotic**: they describe the behavior of the estimator as the sample size $n$ tends to infinity.

## Consistency of the MLE

An estimator $\widehat{\theta}_n$ is called **consistent** for $\theta_0$ if

$$
\widehat{\theta}_n
\xrightarrow{P}
\theta_0.
$$

Equivalently, for every $\varepsilon>0$,

$$
\mathbb P_{\theta_0}
\left(
\left\|
\widehat{\theta}_n-\theta_0
\right\|>\varepsilon
\right)
\longrightarrow 0.
$$

Thus, consistency means that the estimator becomes arbitrarily close to the true parameter with high probability as the sample size increases.

## Consistency Theorem for the MLE

Let

$$
X_1,\ldots,X_n
\overset{\mathrm{iid}}{\sim}
f(x;\theta_0),
$$

where $\theta_0$ is the true parameter value.

Under suitable regularity and identifiability conditions,

$$
\boxed{
\widehat{\theta}_{\mathrm{MLE}}
\xrightarrow{P}
\theta_0.
}
$$

The key idea is that the normalized log-likelihood satisfies

$$
\frac1n\ell_n(\theta)
=
\frac1n
\sum_{i=1}^n
\log f(X_i;\theta)
\xrightarrow{P}
\mathbb E_{\theta_0}
\left[
\log f(X;\theta)
\right].
$$

The limiting objective function is maximized at the true parameter $\theta_0$.

Therefore, the maximizer of the empirical log-likelihood converges to the maximizer of its population counterpart.

## Identifiability

A statistical model is called **identifiable** if distinct parameter values correspond to distinct probability distributions.

Formally,

$$
P_{\theta_1}=P_{\theta_2}
\quad\Longrightarrow\quad
\theta_1=\theta_2.
$$

Equivalently, if

$$
f(x;\theta_1)=f(x;\theta_2)
$$

for almost every $x$, then

$$
\theta_1=\theta_2.
$$

Identifiability is necessary for consistent estimation: if two different parameter values generate exactly the same distribution, no estimator can distinguish between them.

## Why Is the Expected Log-Likelihood Maximized at the Truth?

Define the population criterion

$$
Q(\theta)
=
\mathbb E_{\theta_0}
\left[
\log f(X;\theta)
\right].
$$

Then

$$
Q(\theta_0)-Q(\theta)
=
\mathbb E_{\theta_0}
\left[
\log
\frac{f(X;\theta_0)}
{f(X;\theta)}
\right].
$$

The right-hand side is the Kullback–Leibler divergence

$$
D_{\mathrm{KL}}
\left(
P_{\theta_0}\,\|\,P_\theta
\right).
$$

Since

$$
D_{\mathrm{KL}}
\left(
P_{\theta_0}\,\|\,P_\theta
\right)
\geq 0,
$$

we obtain

$$
Q(\theta_0)\geq Q(\theta).
$$

Under identifiability, equality holds only when

$$
\theta=\theta_0.
$$

Hence, the expected log-likelihood is uniquely maximized at the true parameter.

## Asymptotic Normality of the MLE

Consistency alone states that the estimator converges to the true parameter.

To perform inference, we also need to know the rate of convergence and the approximate distribution of the estimation error.

Under suitable regularity conditions,

$$
\boxed{
\sqrt n
\left(
\widehat{\theta}_{\mathrm{MLE}}-\theta_0
\right)
\xrightarrow{d}
N
\left(
0,
I_1(\theta_0)^{-1}
\right),
}
$$

where $I_1(\theta_0)$ is the Fisher information contained in one observation.

Equivalently, if

$$
I_n(\theta_0)
=
nI_1(\theta_0),
$$

then for large $n$,

$$
\boxed{
\widehat{\theta}_{\mathrm{MLE}}
\approx
N
\left(
\theta_0,
I_n(\theta_0)^{-1}
\right).
}
$$

## Derivation from the Score Equation

The MLE satisfies the score equation

$$
U_n(\widehat{\theta})=0.
$$

Expand the score around the true parameter $\theta_0$:

$$
0
=
U_n(\theta_0)
+
U_n'(\widetilde{\theta})
\left(
\widehat{\theta}-\theta_0
\right),
$$

where $\widetilde{\theta}$ lies between $\widehat{\theta}$ and $\theta_0$.

Therefore,

$$
\sqrt n
\left(
\widehat{\theta}-\theta_0
\right)
=
-
\left[
\frac1n
U_n'(\widetilde{\theta})
\right]^{-1}
\frac{U_n(\theta_0)}{\sqrt n}.
$$

Now,

$$
\frac{U_n(\theta_0)}{\sqrt n}
\xrightarrow{d}
N
\left(
0,
I_1(\theta_0)
\right)
$$

by the Central Limit Theorem, while

$$
-\frac1n
U_n'(\widetilde{\theta})
\xrightarrow{P}
I_1(\theta_0)
$$

by the Law of Large Numbers and consistency.

Hence,

$$
\sqrt n
\left(
\widehat{\theta}-\theta_0
\right)
\xrightarrow{d}
N
\left(
0,
I_1(\theta_0)^{-1}
\right).
$$

## Vector-Valued Parameters

If

$$
\theta\in\mathbb R^p,
$$

then the asymptotic distribution becomes multivariate:

$$
\boxed{
\sqrt n
\left(
\widehat{\theta}_{\mathrm{MLE}}-\theta_0
\right)
\xrightarrow{d}
N_p
\left(
0,
I_1(\theta_0)^{-1}
\right).
}
$$

Here,

$$
I_1(\theta_0)
$$

is the Fisher information matrix for one observation.

Thus, for large samples,

$$
\widehat{\theta}_{\mathrm{MLE}}
\approx
N_p
\left(
\theta_0,
\frac1n
I_1(\theta_0)^{-1}
\right).
$$

## Estimated Asymptotic Variance

The true parameter $\theta_0$ is unknown, so the Fisher information is usually evaluated at the MLE.

Thus,

$$
\operatorname{Avar}
\left(
\widehat{\theta}_{\mathrm{MLE}}
\right)
\approx
I_n
\left(
\widehat{\theta}_{\mathrm{MLE}}
\right)^{-1}.
$$

Alternatively, one may use the observed Fisher information,

$$
J_n
\left(
\widehat{\theta}_{\mathrm{MLE}}
\right)
=
-
\ell_n''
\left(
\widehat{\theta}_{\mathrm{MLE}}
\right).
$$

Therefore,

$$
\boxed{
\widehat{\operatorname{Var}}
\left(
\widehat{\theta}_{\mathrm{MLE}}
\right)
=
J_n
\left(
\widehat{\theta}_{\mathrm{MLE}}
\right)^{-1}.
}
$$

The square root of this quantity is called the **estimated standard error** of the MLE.

## Asymptotic Efficiency

An estimator is called **asymptotically efficient** if its asymptotic variance attains the smallest possible variance among regular estimators.

For a scalar parameter, the Cramér–Rao Lower Bound suggests that the smallest possible asymptotic variance is

$$
\frac{1}{nI_1(\theta_0)}.
$$

Since

$$
\widehat{\theta}_{\mathrm{MLE}}
\approx
N
\left(
\theta_0,
\frac{1}{nI_1(\theta_0)}
\right),
$$

the MLE attains this lower bound asymptotically.

Therefore,

$$
\boxed{
\operatorname{Avar}
\left(
\widehat{\theta}_{\mathrm{MLE}}
\right)
=
\frac{1}{nI_1(\theta_0)}.
}
$$

In this sense, the MLE is asymptotically efficient.

## Important Remark on Efficiency

The MLE need not be unbiased for finite $n$.

It may also fail to attain the Cramér–Rao Lower Bound exactly in small samples.

Asymptotic efficiency means that, as $n\to\infty$,

$$
n\operatorname{Var}
\left(
\widehat{\theta}_{\mathrm{MLE}}
\right)
\longrightarrow
I_1(\theta_0)^{-1}.
$$

Thus, the efficiency of the MLE is primarily a large-sample property.

## Invariance Property of the MLE

One of the most useful properties of maximum likelihood estimation is its invariance under transformations.

Suppose that

$$
\widehat{\theta}_{\mathrm{MLE}}
$$

is the MLE of $\theta$, and let

$$
\eta=g(\theta)
$$

be a transformation of the parameter.

Then the MLE of $\eta$ is

$$
\boxed{
\widehat{\eta}_{\mathrm{MLE}}
=
g
\left(
\widehat{\theta}_{\mathrm{MLE}}
\right).
}
$$

Thus, to estimate a function of the parameter, one simply applies the same function to the MLE.

## Invariance Theorem

Let

$$
\widehat{\theta}
\in
\operatorname*{arg\,max}_{\theta\in\Theta}
L(\theta).
$$

Suppose that

$$
\eta=g(\theta).
$$

Then an MLE of $\eta$ is

$$
\widehat{\eta}
=
g(\widehat{\theta}).
$$

If $g$ is one-to-one, the result follows immediately by reparameterizing the likelihood.

If $g$ is not one-to-one, the profile likelihood for $\eta$ is defined by

$$
L^*(\eta)
=
\sup_{\theta:g(\theta)=\eta}
L(\theta).
$$

Since $\widehat{\theta}$ maximizes $L(\theta)$, the value

$$
g(\widehat{\theta})
$$

maximizes $L^*(\eta)$.

## Example of Invariance

Suppose that

$$
X_1,\ldots,X_n
\overset{\mathrm{iid}}{\sim}
N(\mu,\sigma^2),
$$

and that the MLE of $\sigma^2$ is

$$
\widehat{\sigma}^2_{\mathrm{MLE}}
=
\frac1n
\sum_{i=1}^n
(X_i-\overline X)^2.
$$

Since

$$
\sigma
=
\sqrt{\sigma^2},
$$

the invariance property implies that the MLE of $\sigma$ is

$$
\boxed{
\widehat{\sigma}_{\mathrm{MLE}}
=
\sqrt{
\frac1n
\sum_{i=1}^n
(X_i-\overline X)^2
}.
}
$$

No separate maximization is required.

## Summary of the Main MLE Properties

Under suitable regularity conditions, the maximum likelihood estimator satisfies

$$
\boxed{
\widehat{\theta}_{\mathrm{MLE}}
\xrightarrow{P}
\theta_0
}
$$

and

$$
\boxed{
\sqrt n
\left(
\widehat{\theta}_{\mathrm{MLE}}-\theta_0
\right)
\xrightarrow{d}
N
\left(
0,
I_1(\theta_0)^{-1}
\right).
}
$$

Consequently, the MLE is asymptotically efficient.

Moreover, for any transformation

$$
\eta=g(\theta),
$$

the corresponding MLE is

$$
\boxed{
\widehat{\eta}_{\mathrm{MLE}}
=
g
\left(
\widehat{\theta}_{\mathrm{MLE}}
\right).
}
$$

These properties can be summarized as follows:

| Property | Mathematical statement |
|---|---|
| Consistency | $\widehat{\theta}_{\mathrm{MLE}}\xrightarrow{P}\theta_0$ |
| Asymptotic normality | $\sqrt n(\widehat{\theta}_{\mathrm{MLE}}-\theta_0)\xrightarrow{d}N(0,I_1(\theta_0)^{-1})$ |
| Asymptotic efficiency | $\operatorname{Avar}(\widehat{\theta}_{\mathrm{MLE}})=I_n(\theta_0)^{-1}$ |
| Invariance | $\widehat{g(\theta)}_{\mathrm{MLE}}=g(\widehat{\theta}_{\mathrm{MLE}})$ |

# Quantile Estimation

In many applications, the parameter of interest is not the mean or the variance of a distribution, but one of its **quantiles**.

Examples include

- the median household income,
- the 95th percentile of exam scores,
- the Value-at-Risk (VaR) in finance,
- and confidence limits in quality control.

Unlike moment estimation, quantile estimation relies on the order of the observations rather than on their numerical averages.

The theory of quantile estimation is therefore closely connected with **order statistics**.

## Population Quantiles

Let $X$ be a continuous random variable with cumulative distribution function

$$
F(x).
$$

### Definition

For a probability level

$$
0<p<1,
$$

the **$p$-th population quantile** is the value

$$
\boxed{
q_p
=
F^{-1}(p),
}
$$

that satisfies

$$
\boxed{
F(q_p)=p.
}
$$

Equivalently,

$$
\boxed{
P(X\le q_p)=p.
}
$$

Thus, a proportion $p$ of the population lies below the quantile $q_p$.

## Examples

Some commonly used quantiles are

| Probability | Quantile | Interpretation |
|-------------|----------|----------------|
| $0.25$ | First quartile ($Q_1$) | 25% of observations lie below |
| $0.50$ | Median | Half of the observations lie below |
| $0.75$ | Third quartile ($Q_3$) | 75% of observations lie below |
| $0.90$ | 90th percentile | Top 10% begin above this value |
| $0.95$ | 95th percentile | Frequently used in risk analysis |
| $0.99$ | 99th percentile | Extreme observations |

## Quantiles and the Distribution Function

Since the cumulative distribution function is increasing,

$$
F^{-1}(p)
$$

exists whenever $F$ is strictly increasing.

Consequently,

$$
\boxed{
q_p
=
\inf
\left\{
x:
F(x)\ge p
\right\}.
}
$$

This definition also applies to discrete distributions, where the inverse function may not exist in the usual sense.

# Sample Quantiles

Suppose

$$
X_1,\ldots,X_n
$$

is a random sample.

Arrange the observations in increasing order,

$$
X_{(1)}
\le
X_{(2)}
\le
\cdots
\le
X_{(n)},
$$

where

$$
X_{(1)},\ldots,X_{(n)}
$$

are the **order statistics**.

The sample quantiles are constructed from these ordered observations.

## Definition

For

$$
0<p<1,
$$

let

$$
k
=
\lfloor
(n+1)p
\rfloor.
$$

The **sample $p$-th quantile** is defined approximately by

$$
\boxed{
\widehat q_p
=
X_{(k)}.
}
$$

In practice, many statistical software packages interpolate between neighboring order statistics.

Different interpolation rules produce slightly different finite-sample estimators but are asymptotically equivalent.

## Interpretation

The sample quantile is the empirical analogue of the population quantile.

Indeed,

$$
\boxed{
\widehat q_p
\approx
q_p.
}
$$

Thus,

approximately

- a proportion $p$ of the observations lie below

$$
\widehat q_p,
$$

- while a proportion

$$
1-p
$$

lie above it.

# The Sample Median

The most important quantile is the median,

$$
q_{0.5}.
$$

The sample median is therefore the estimator of the population median.

## Definition

Suppose the observations are ordered,

$$
X_{(1)}
\le
\cdots
\le
X_{(n)}.
$$

If

$$
n
$$

is odd,

$$
\boxed{
\widetilde X
=
X_{\left(\frac{n+1}{2}\right)}.
}
$$

If

$$
n
$$

is even,

$$
\boxed{
\widetilde X
=
\frac{
X_{(n/2)}
+
X_{(n/2+1)}
}{2}.
}
$$

The statistic

$$
\widetilde X
$$

is called the **sample median**.

## Why Is the Median Important?

Unlike the sample mean, the sample median depends only on the order of the observations.

Consequently,

- it is much less sensitive to outliers,

- it provides a robust measure of central tendency,

- and it exists whenever the observations can be ordered.

For heavily skewed distributions, the median often provides a more representative measure of location than the mean.

# Consistency of Sample Quantiles

Suppose that

$$
F
$$

is continuous and strictly increasing.

Then

$$
\boxed{
\widehat q_p
\xrightarrow{P}
q_p.
}
$$

Thus, sample quantiles are consistent estimators of the corresponding population quantiles.

## Asymptotic Distribution of Sample Quantiles

Consistency alone does not describe the sampling variability.

Under suitable regularity conditions,

$$
\boxed{
\sqrt n
\left(
\widehat q_p-q_p
\right)
\xrightarrow{d}
N
\left(
0,
\frac{
p(1-p)
}{
f(q_p)^2
}
\right),
}
$$

where

$$
f(q_p)
$$

is the value of the probability density at the true quantile.

## Interpretation

The asymptotic variance

$$
\boxed{
\frac{
p(1-p)
}{
f(q_p)^2
}
}
$$

depends on two factors.

First,

$$
p(1-p)
$$

depends only on the probability level.

Second,

$$
f(q_p)
$$

depends on the local shape of the probability density.

If the density is large near the quantile, the observations are highly concentrated around the true quantile, leading to a more precise estimator.

Conversely, if the density is small, the observations are more dispersed and the quantile estimator becomes less accurate.

# Asymptotic Distribution of the Sample Median

The sample median corresponds to

$$
p=\frac12.
$$

Substituting

$$
p=\frac12
$$

into the previous theorem gives

$$
\boxed{
\sqrt n
\left(
\widetilde X-m
\right)
\xrightarrow{d}
N
\left(
0,
\frac{
1
}{
4f(m)^2
}
\right),
}
$$

where

$$
m
$$

denotes the population median.

## Comparison with the Sample Mean

For a normal population,

$$
X
\sim
N(\mu,\sigma^2),
$$

the asymptotic variances are

$$
\operatorname{Var}
(\overline X)
\approx
\frac{\sigma^2}{n},
$$

and

$$
\operatorname{Var}
(\widetilde X)
\approx
\frac{\pi}{2}
\cdot
\frac{\sigma^2}{n}.
$$

Hence,

$$
\boxed{
\operatorname{Var}
(\widetilde X)
\approx
1.57
\,
\operatorname{Var}
(\overline X).
}
$$

Therefore, under normality, the sample mean is more efficient than the sample median.

However, the sample median is substantially more robust in the presence of outliers or heavy-tailed distributions.

and

$$

The sample median is the special case

$$

$$

and is one of the most widely used robust estimators in statistics.

# A Unified View of Estimation

Although the Method of Moments and Maximum Likelihood Estimation appear to be fundamentally different procedures, they are both examples of a more general principle.

Rather than searching directly for an estimator, we construct an equation involving the unknown parameter and the observed sample.

The estimator is then defined as the solution of this equation.

Such equations are called **estimating equations**.

This viewpoint provides a unified framework for many estimation methods developed in modern statistics and econometrics.

## Estimating Equations

Let

$$
X_1,\ldots,X_n
$$

be a random sample from a distribution depending on an unknown parameter

$$
\theta.
$$

Suppose that

$$
\psi(x,\theta)
$$

is a vector-valued function satisfying

$$
\boxed{
\mathbb E_\theta
\left[
\psi(X,\theta)
\right]
=
0.
}
$$

The estimator is obtained by solving the empirical equation

$$
\boxed{
\frac1n
\sum_{i=1}^{n}
\psi(X_i,\theta)
=
0.
}
$$

This equation is called an **estimating equation**.

The resulting solution is called an **estimating equation estimator**.

## The Method of Moments as an Estimating Equation

Suppose the unknown parameter

$$
\theta
$$

is determined by the first

$$
r
$$

population moments.

Define

$$
\psi(X,\theta)
=
\begin{pmatrix}
X-\mu_1'(\theta)
\\
X^2-\mu_2'(\theta)
\\
\vdots
\\
X^r-\mu_r'(\theta)
\end{pmatrix}.
$$

Then

$$
\mathbb E_\theta
[
\psi(X,\theta)
]
=
0.
$$

The empirical estimating equation becomes

$$
\boxed{
\frac1n
\sum_{i=1}^{n}
\psi(X_i,\theta)
=
0,
}
$$

which is precisely the Method of Moments.

Thus, the Method of Moments estimates the parameter by forcing the empirical moments to equal their theoretical counterparts.

## Maximum Likelihood as an Estimating Equation

Recall that the score function is

$$
U(\theta)
=
\frac{\partial}
{\partial\theta}
\ell(\theta).
$$

The maximum likelihood estimator satisfies

$$
\boxed{
U(\widehat\theta)=0.
}
$$

Since

$$
\ell(\theta)
=
\sum_{i=1}^{n}
\log f(X_i;\theta),
$$

the score may be written as

$$
U(\theta)
=
\sum_{i=1}^{n}
\frac{\partial}
{\partial\theta}
\log f(X_i;\theta).
$$

Dividing by

$$
n,
$$

we obtain

$$
\boxed{
\frac1n
\sum_{i=1}^{n}
\frac{\partial}
{\partial\theta}
\log f(X_i;\theta)
=
0.
}
$$

Therefore, Maximum Likelihood Estimation is also an estimating equation method.

The estimating function is simply

$$
\boxed{
\psi(X,\theta)
=
\frac{\partial}
{\partial\theta}
\log f(X;\theta).
}
$$

## Why Does This Matter?

Viewing estimation through estimating equations provides a unified perspective.

Both estimation methods consist of finding the value of the parameter that satisfies an empirical equation,

$$
\frac1n
\sum_{i=1}^{n}
\psi(X_i,\theta)
=
0.
$$

The only difference lies in the choice of the estimating function

$$
\psi.
$$

This idea forms the foundation of many modern estimation procedures, including

- Generalized Method of Moments (GMM),
- M-estimation,
- Generalized Estimating Equations (GEE),
- and Quasi-Likelihood methods.

# Comparison of Estimation Methods

The Method of Moments and Maximum Likelihood Estimation are the two most widely used general estimation procedures.

Although both produce consistent and asymptotically normal estimators under suitable assumptions, they differ substantially in the amount of information they use and in their statistical efficiency.

## Method of Moments vs. Maximum Likelihood

| Property | Method of Moments | Maximum Likelihood |
|:---|:---:|:---:|
| Uses moments only | ✓ | ✗ |
| Uses the full probability model | ✗ | ✓ |
| Requires optimization | Usually not | Usually yes |
| Closed-form estimators | Often | Sometimes |
| Consistent | ✓ | ✓ |
| Asymptotically normal | ✓ | ✓ |
| Asymptotically efficient | Usually ✗ | ✓ |
| Invariant under transformations | ✗ | ✓ |
| Computational complexity | Low | Moderate to high |

## When Should Each Method Be Used?

The Method of Moments is often preferable when

- the likelihood is difficult to derive,

- computational simplicity is important,

- or a quick preliminary estimate is required.

Maximum Likelihood Estimation is generally preferred when

- the probability model is fully specified,

- accurate statistical inference is required,

- standard errors and confidence intervals are needed,

- or hypothesis testing is to be performed.

In many practical applications, Method of Moments estimators are used as initial values for numerical algorithms that compute the Maximum Likelihood Estimator.

## Main Ideas of This Chapter

This chapter introduced two of the most important estimation methods in mathematical statistics.

The **Method of Moments** estimates unknown parameters by matching theoretical and empirical moments.

The **Maximum Likelihood Method** estimates the parameter that makes the observed sample most likely under the assumed statistical model.

Although these methods are constructed differently, they share many desirable asymptotic properties and can both be viewed as solving estimating equations.

This unified perspective extends naturally to many advanced estimation methods encountered in modern statistics, econometrics, and machine learning.

Suppose that

$$
X_1,\ldots,X_n
$$

is a random sample from a distribution depending on an unknown parameter

$$
\theta.
$$

The two principal estimation methods studied in this chapter are summarized below.

| Method | Fundamental Idea | Estimating Equation |
|:---|:---|:---|

- both estimators are consistent,

- both are asymptotically normal,

For this reason, maximum likelihood estimation has become the standard estimation procedure in modern statistical inference.
