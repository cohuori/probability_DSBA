During the seminars, we will work on problems of three different difficulty levels. **Category 1** (<span style="color:#28a745;"><strong>green</strong></span>) contains introductory problems, **Category 2** (<span style="color:#007bff;"><strong>blue</strong></span>) contains standard problems, and **Category 3** (<span style="color:#fd7e14;"><strong>orange</strong></span>) contains the most challenging problems.


# Category 1

::: {.callout-tip collapse="true"}
## Problem 1

Cars independently pass a point on a busy road at an average rate of 150 per hour.

1. Assuming a Poisson distribution, find the probability that none passes in one minute.
2. What is the expected number of cars passing in two minutes?
3. Find the probability that exactly the expected number of cars passes in a two-minute period.

Other motor vehicles (vans, motorcycles, etc.) pass the same point independently at an average rate of 75 per hour.

4. Find the probability that exactly one car and one other motor vehicle pass during a two-minute period.
:::

::: {.callout-tip collapse="true"}
## Problem 2

Telephone calls enter a hotel switchboard according to a Poisson process at an average rate of two every three minutes.

Let $Y$ denote the waiting time until the second call.

1. Find the probability that exactly one call arrives during three minutes.
2. Find the probability that no more than one call arrives during three minutes.
3. Compute $P(Y>3)$.
:::

::: {.callout-tip collapse="true"}

## Problem 3

Suppose that the heights (in cm) of Russian adults satisfy

- Men: $X\sim N(174.9,7.39^2)$
- Women: $Y\sim N(161.3,6.85^2)$

Assume independence.

Find the probability that a randomly selected man is at most $10$ cm taller than a randomly selected woman.
:::

## Problem 4

Let $X_1,\ldots,X_8$ be i.i.d. $N(0,16)$, $Y_1,\ldots,Y_5$ be i.i.d. $N(1,9)$, $Z_1,\ldots,Z_4$ be i.i.d. $N(0,1)$, all independent.

1. $P\!\left(\sum_{i=1}^{4}Z_i^2<5\right)$

2. Find $k$ such that

$$
P\!\left(\sum_{i=1}^{4}Z_i^2<k\right)=0.4.
$$

3. $P\!\left(\sum_{i=1}^{8}X_i^2<20\right)$

4. $P\!\left(\sum_{i=1}^{5}(Y_i-1)^2>32\right)$
:::

# Category 2

::: {.callout-note collapse="true"}
## Problem 1
Prove the memoryless property of the exponential distribution.

If

$$
X\sim\operatorname{Exp}(\lambda),
$$

show that

$$
P(X>s+t\mid X>s)=P(X>t),\qquad s,t>0.
$$

:::

::: {.callout-note collapse="true"}
## Problem 2
Let $X$ and $Y$ be two jointly continuous random variables with joint PDF:
$$f_{XY}(x,y)=\begin{cases}
            x+cy^2, \quad 0\leq x\leq 1, 0\leq y \leq 1 \\
            0, \quad \quad \quad \text{otherwise}
        \end{cases}$$

1. Find the constant $c$.

2. Find P($0\leq X\leq 0.5, 0\leq Y\leq 0.5$).

:::

## Problem 2
The CDF of a random vector $(\xi, \eta)$ is given by

\begin{equation*}
    F_{\xi,\eta}(x,y) = \left\{
  \begin{array}{lr}
    0 & , x \leq 0 \ \text{or} \ y\leq 0,;\\
    1-e^{-x^2}-e^{-2y}+e^{-x^2-2y} & , x\geq 0, y\geq 0.
  \end{array}
\right.
\end{equation*}

Find

1. The joint PDF $p_{\xi, \eta}(x,y)$.

2. The probabilities ${\sf P}(-2\leq \xi <2, 1\leq \eta <3)$, ${\sf P}(\xi \geq 0, \eta \geq 1)$, ${\sf P}(-2\leq \xi <2, 1\leq \eta <3)$, ${\sf P}( \xi <1, \eta \geq 2)$ 

3. CDFs $F_{\xi}$ and $F_{\eta}$.
:::
