# Category 1

::: {.callout-tip collapse="true"}
## Problem 1
The PDF of $\eta$ is given by

\begin{equation*}
    p_{\eta}(x) = \left\{
  \begin{array}{lr}
    A \sin 2x & , x \in [0,\frac{\pi}{2}];\\
    0 & , \text{otherwise}.
  \end{array}
\right.
\end{equation*}

Find (a) Value of $A$. (b)  CDF of $\eta$, (c) ${\sf E} \eta$, ${\sf Var} \eta$. (d) ${\sf P}(-\frac{\pi}{6}\leq \eta \leq \frac{\pi}{6})$.

:::

::: {.callout-tip collapse="true"}
## Problem 2
The probability distribution function (PDF) of a random vector $(\xi,\eta)$ is given by

\begin{equation*}
    p_{\xi,\eta}(x,y) = \left\{
  \begin{array}{lr}
    Axy & , (x,y) \in D;\\
    0 & ,(x,y) \notin D.
  \end{array}
\right.
\end{equation*}

where $D:\{x\geq0,y\geq 0, x+y\leq 2 \}$. Find 

1. The value of $A$.

2. ${\sf P}((\xi,\eta) \in G)$, where $G:\{0\leq x\leq 1, 0\leq y\leq 1  \}$.

3. $p_{\xi}(x)$, $p_{\eta}(y)$. Are $\xi$ and $\eta$ independent random variables?

4. ${\sf E} \xi$, ${\sf E} \eta$.

5. ${\sf Var} \xi$, ${\sf Var} \eta$.

6. ${\sf Cov} (\xi,\eta)$ and ${\sf Corr} (\xi,\eta)$

:::

::: {.callout-tip collapse="true"}
## Problem 3
 Let $X$ and $Y$ be two jointly continuous random variables with joint PDF:
$$f_{XY}(x,y)=\begin{cases}
            cxy^2, \quad 0\leq y\leq x \leq 1 \\
            0, \quad \quad \quad \text{otherwise}
        \end{cases}$$
        
1. Find th constant $c$.

2. Find $R_{XY}$ and show it in the $xy$ plane.

3. Find marginal PDFs, $f_X(x$) and $f_Y(y)$.

4. Find P$\left(Y\leq \dfrac{X}{2}\right)$.

5. Find P$\left(Y\leq \dfrac{X}{4}\bigg|Y\leq \dfrac{X}{2}\right)$.

:::

::: {.callout-tip collapse="true"}
## Problem 4
The joint probability density function of the random variables $X$ and $Y$ is given by
\begin{equation*}
    p_{X,Y}(x,y) = \left\{
  \begin{array}{lr}
    \frac{3}{2}(x^2 + y^2) & , 0 \leq x \leq 1,\; 0 \leq y \leq 1;\\
    0 & , \text{otherwise.}
  \end{array}
\right.
\end{equation*}
Find ${\sf cov}(X,Y)$ and ${\sf corr}(X,Y)$.

:::

::: {.callout-tip collapse="true"}
## Problem 5
The joint probability density function of the random variables $X$ and $Y$ is given by
\begin{equation*}
    p_{X,Y}(x,y) = \left\{
  \begin{array}{lr}
    4xy & , 0 \leq x \leq 1, \; 0 \leq y \leq 1;\\
    0 & , \text{otherwise.}
  \end{array}
\right.
\end{equation*}
Find ${\sf E}X^2Y$ and ${\sf cov}(X,Y)$.
